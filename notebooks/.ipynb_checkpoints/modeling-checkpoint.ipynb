```python
# ── Run every session ──────────────────────────────────────────────
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import joblib
import os
import shap
import warnings
warnings.filterwarnings('ignore')
from sklearn.ensemble import RandomForestClassifier
from sklearn.model_selection import cross_val_score

# Reload feature matrix
df = pd.read_csv('../data/feature_matrix.csv')

# Reload feature columns
feature_cols = joblib.load('../models/feature_columns.pkl')
X = df[feature_cols]

# Reload all trained disease models
trained_diseases = joblib.load('../models/trained_diseases.pkl')
confidence_tiers = joblib.load('../models/confidence_tiers.pkl')

# Reload training report
report_df = pd.read_csv('../data/training_report.csv')

disease_models = {}
for disease in trained_diseases:
    safe_name = disease.replace(' ', '_').replace('/', '_').replace("'", '')
    path = f'../models/disease_models/{safe_name}.pkl'
    if os.path.exists(path):
        disease_models[disease] = joblib.load(path)

print(f'Feature matrix loaded: {df.shape}')
print(f'Feature columns: {feature_cols}')
print(f'Disease models reloaded: {len(disease_models)}')
print('Ready to continue')

# ── Reload prediction function ─────────────────────────────────────
def predict_diseases(user_features: dict, top_n=5):
    input_df = pd.DataFrame([user_features])[feature_cols]
    scores = {}
    for disease, model in disease_models.items():
        prob_positive = model.predict_proba(input_df)[0][1]
        scores[disease] = prob_positive
    ranked = sorted(scores.items(), key=lambda x: x[1], reverse=True)
    return [(disease, round(prob * 100, 1)) for disease, prob in ranked[:top_n]]

print('Prediction function ready')
```

```python
import os
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
import joblib
from sklearn.ensemble import RandomForestClassifier
from sklearn.naive_bayes import GaussianNB
from sklearn.model_selection import train_test_split, cross_val_score
from sklearn.metrics import classification_report, accuracy_score, confusion_matrix
from sklearn.preprocessing import LabelEncoder
import shap
import warnings
warnings.filterwarnings('ignore')

df = pd.read_csv('../data/feature_matrix.csv')
print(df.shape)
print(df.head())
```

```python
# Quick audit of positive/negative rows per disease
summary = df.groupby('Disease')['Outcome Variable'].agg(
    total='count',
    positive='sum',
    negative=lambda x: (x == 0).sum()
).reset_index()

print('=== Dataset Audit ===')
print(f'Total diseases: {df["Disease"].nunique()}')
print(f'Diseases with at least 1 positive row: {(summary["positive"] > 0).sum()}')
print(f'Diseases with zero positive rows: {(summary["positive"] == 0).sum()}')
print(f'Total rows: {len(df)}')
print(f'Total positive rows: {(df["Outcome Variable"] == 1).sum()}')
print(f'Total negative rows: {(df["Outcome Variable"] == 0).sum()}')
print()
print('Diseases with NO positive rows (cannot be predicted):')
no_positive = summary[summary['positive'] == 0]['Disease'].tolist()
for d in no_positive:
    print(f'  - {d}')


summary.to_csv('../data/disease_audit.csv', index=False)
print('Disease audit saved')
```

Problem: over half of the diseases only have 1 row, so we can't do a train/test split. Further, 39 of the diseases have no positive rows, so a model can't predict what the disease looks like. Because of these limitations, we will use one-vs-rest binary classification, building labels that indicate confirmed results strictly when there is a positive case of the disease we are currently training.

```python
feature_cols = [
    'Fever',
    'Cough',
    'Fatigue',
    'Difficulty Breathing',
    'Age',
    'Gender',
    'Blood Pressure',
    'Cholesterol Level'
]

# X does NOT include Outcome Variable or Disease
# Outcome Variable is used only for labelling, not as a feature
X = df[feature_cols]

print('Feature columns:', feature_cols)
print('Feature matrix shape:', X.shape)
```

```python
os.makedirs('../models/disease_models', exist_ok=True)

disease_models = {}
training_report = []

all_diseases = df['Disease'].unique()

for disease in all_diseases:
    # Label = 1 only when it IS this disease AND confirmed positive
    labels = ((df['Disease'] == disease) & (df['Outcome Variable'] == 1)).astype(int)

    positive_count = labels.sum()
    negative_count = (labels == 0).sum()

    # Skip diseases with no positive examples — nothing to learn
    if positive_count == 0:
        training_report.append({
            'disease': disease,
            'status': 'skipped',
            'positive_rows': 0,
            'negative_rows': negative_count
        })
        continue

    model = RandomForestClassifier(
        n_estimators=200,
        max_depth=None,       # let trees grow fully for small datasets
        min_samples_leaf=1,
        class_weight='balanced',  # critical — most rows are 0, this rebalances
        random_state=42
    )
    model.fit(X, labels)

    # Save the model
    safe_name = disease.replace(' ', '_').replace('/', '_').replace("'", '')
    joblib.dump(model, f'../models/disease_models/{safe_name}.pkl')

    disease_models[disease] = model
    training_report.append({
        'disease': disease,
        'status': 'trained',
        'positive_rows': int(positive_count),
        'negative_rows': int(negative_count)
    })

report_df = pd.DataFrame(training_report)
print(f'Successfully trained: {len(disease_models)} models')
print(f'Skipped (no positive rows): {(report_df["status"] == "skipped").sum()}')
print()
print('Trained diseases:')
print(report_df[report_df['status'] == 'trained'][['disease', 'positive_rows', 'negative_rows']])
```

```python
# Save feature column order — critical for consistent predictions in the app
joblib.dump(feature_cols, '../models/feature_columns.pkl')

# Save list of trained diseases — the app will only show these as possible outputs
trained_diseases = list(disease_models.keys())
joblib.dump(trained_diseases, '../models/trained_diseases.pkl')

# Save the full training report for reference
report_df.to_csv('../data/training_report.csv', index=False)

print(f'Saved {len(trained_diseases)} disease models')
print('Feature columns saved')
print('Training report saved to data/training_report.csv')
```

```python
def predict_diseases(user_features: dict, top_n=5):
    """
    Takes a dictionary of user inputs (no Outcome Variable needed),
    runs all disease models, and returns the top N predictions
    ranked by confidence.

    Parameters:
        user_features: dict with keys matching feature_cols
        top_n: how many predictions to return

    Returns:
        list of (disease_name, confidence_percentage) tuples
    """
    input_df = pd.DataFrame([user_features])[feature_cols]

    scores = {}
    for disease, model in disease_models.items():
        # predict_proba returns [prob_class_0, prob_class_1]
        # we want the probability of class 1 (disease confirmed)
        prob_positive = model.predict_proba(input_df)[0][1]
        scores[disease] = prob_positive

    # Sort by probability descending and take top N
    ranked = sorted(scores.items(), key=lambda x: x[1], reverse=True)

    return [(disease, round(prob * 100, 1)) for disease, prob in ranked[:top_n]]
```

Prediction function tests:

```python
# Test 1 — classic flu symptoms
print('=== Test 1: Flu-like symptoms ===')
sample1 = {
    'Fever': 1,
    'Cough': 1,
    'Fatigue': 1,
    'Difficulty Breathing': 0,
    'Age': 35,
    'Gender': 1,
    'Blood Pressure': 1,
    'Cholesterol Level': 1
}
for disease, confidence in predict_diseases(sample1):
    print(f'  {disease}: {confidence}%')

print()

# Test 2 — respiratory symptoms, older patient, high blood pressure
print('=== Test 2: Respiratory + older patient ===')
sample2 = {
    'Fever': 0,
    'Cough': 1,
    'Fatigue': 1,
    'Difficulty Breathing': 1,
    'Age': 65,
    'Gender': 0,
    'Blood Pressure': 2,
    'Cholesterol Level': 2
}
for disease, confidence in predict_diseases(sample2):
    print(f'  {disease}: {confidence}%')

print()

# Test 3 — minimal symptoms, young patient
print('=== Test 3: Minimal symptoms, young patient ===')
sample3 = {
    'Fever': 0,
    'Cough': 0,
    'Fatigue': 1,
    'Difficulty Breathing': 0,
    'Age': 22,
    'Gender': 1,
    'Blood Pressure': 1,
    'Cholesterol Level': 0
}
for disease, confidence in predict_diseases(sample3):
    print(f'  {disease}: {confidence}%')
```

Different inputs produce different results, which is indicative of a working model. Obvious pitfalls include lack of symptoms/patient profile statistics to predict diseases from, leading to inaccurate results (many diseases that have the exact same symptoms and patient profiles can turn positive). Accuracy will improve if/when more data is incorporated into the model.

```python
print('=== Cross Validation Per Disease ===\n')

cv_results = []

for disease in list(disease_models.keys())[:10]:  # evaluate top 10 for speed
    labels = ((df['Disease'] == disease) & (df['Outcome Variable'] == 1)).astype(int)

    model = RandomForestClassifier(
        n_estimators=200,
        class_weight='balanced',
        random_state=42
    )

    # 3-fold cross validation — splits data 3 ways and tests on each
    scores = cross_val_score(model, X, labels, cv=3, scoring='roc_auc')

    cv_results.append({
        'disease': disease,
        'mean_auc': round(scores.mean(), 3),
        'std_auc': round(scores.std(), 3),
        'positive_rows': int(labels.sum())
    })
    print(f'{disease}: AUC = {scores.mean():.3f} (+/- {scores.std():.3f})')

cv_df = pd.DataFrame(cv_results)
print(f'\nAverage AUC across evaluated diseases: {cv_df["mean_auc"].mean():.3f}')
```

Using train/test evaluation and accuracy in determining the utility of the model wouldn't be appropriate; because the negatives far outweigh the positives for each disease in the labels, the model could predict negative every time and have high accuracy. Instead, using AUC will measure how well the model separates positives from negatives. We are aiming for above .7. Further, we are using 3-fold cross-validation, as testing the model on different splits of the data will give us a better sense of how accurate the model is.

It would be optimal to use higher folds, but out dataset is limited in the number of positives (correct disease and positive outcome) it will contain per disease, we cannot use higher folds without sacrificing the possibility that many of our AUC scores will return *nan*. This is due to the fact that AUC can't give a result when there are no positives, because then it wouldn't be able to know how accurately it can separate positives from negatives. There are already a few *nan* results above, but using less folds and in turn creating more abundant yet less accurate results would not be relevant to this project.

AUC of .611 is not great, but to be expected given the limitations of our data. It is doing better than randomly guessing (.5).

We only want to evaluate diseases that have enough data; so, we will remove the *nan* diseases (the ones that do not have enough positives to return results for all the splits). To do this, we will use StratifiedKFold, ensuring that positives are spread across all folds so that each fold has 1 and can compute an AUC. We will also remove diseases that don't have 3 or more positives.

```python
from sklearn.model_selection import StratifiedKFold

cv_results = []

for disease in list(disease_models.keys())[:10]:
    labels = ((df['Disease'] == disease) & (df['Outcome Variable'] == 1)).astype(int)

    # Skip if too few positive examples for 3 fold CV
    if labels.sum() < 3:
        print(f'{disease}: skipped — only {labels.sum()} positive rows, too few for 3-fold CV')
        continue

    model = RandomForestClassifier(
        n_estimators=200,
        class_weight='balanced',
        random_state=42
    )

    # StratifiedKFold ensures each fold has positive examples
    cv = StratifiedKFold(n_splits=3, shuffle=True, random_state=42)
    scores = cross_val_score(model, X, labels, cv=cv, scoring='roc_auc')

    cv_results.append({
        'disease': disease,
        'mean_auc': round(scores.mean(), 3),
        'std_auc': round(scores.std(), 3),
        'positive_rows': int(labels.sum())
    })
    print(f'{disease}: AUC = {scores.mean():.3f} (+/- {scores.std():.3f})')
```

```python
cv_df = pd.DataFrame(cv_results)
print(f'\nAverage AUC across evaluated diseases: {cv_df["mean_auc"].mean():.3f}')
```

Many diseases have very few examples of train on. The model will be less confident in turn. It will be useful to gauge the confidence of the model based on the number of examples it has:

```python
def get_confidence_tier(positive_rows):
    if positive_rows >= 10:
        return 'High'
    elif positive_rows >= 5:
        return 'Medium'
    elif positive_rows >= 2:
        return 'Low'
    else:
        return 'Very Low'

confidence_tiers = {}
for _, row in report_df[report_df['status'] == 'trained'].iterrows():
    confidence_tiers[row['disease']] = get_confidence_tier(row['positive_rows'])

joblib.dump(confidence_tiers, '../models/confidence_tiers.pkl')

print('Confidence tiers:')
for tier in ['High', 'Medium', 'Low', 'Very Low']:
    diseases_in_tier = [d for d, t in confidence_tiers.items() if t == tier]
    print(f'  {tier}: {len(diseases_in_tier)} diseases')
```
